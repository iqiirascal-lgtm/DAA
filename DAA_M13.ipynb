```python
# Implementasi Algoritma Breadth-first Search

graph = {'Amin' : ['Wasim', 'Nick', 'Mike'],
         'Wasim' : ['Imran', 'Amin'],
         'Imran' : ['Wasim', 'Faras'],
         'Faras' : ['Imran'],
         'Mike' : ['Amin',],
         'Nick' : ['Amin'],
         }
```

```python
def bfs(graph, start):
    visited = []
    queue = []
    queue = [start]
    while queue:
      node = queue.pop(0)
      if node not in visited:
        visited.append(node)
        neighbours = graph[node]
        for neighbour in neighbours:
          queue.append(neighbour)
    return visited
```

```python
# IMPLEMENTASI ALGORITMA BREADTH-FIRST SEARCH

# O Amin -------------------
# |           |            |
# O Wasim   O Nick       O Mike
# |
# O Imran
# |
# O Faras

bfs(graph, 'Amin')
```

```python
# O Amin -------------------
# |           |            |
# O Wasim   O Nick       O Mike
# |
# O Imran
# |
# O Faras

bfs(graph, 'Wasim')
```

```python
# Latihan 1

from collections import deque

graph = {
    "Rektor": ["Warek 1", "Warek 2"],
    "Warek 1": ["Kaprodi 1"],
    "Warek 2": ["Kaprodi 2", "Kaprodi 3"],
    "Kaprodi 1": ["Dosen A"],
    "Kaprodi 2": ["Dosen D", "Dosen E"],
    "Kaprodi 3": ["Dosen F", "Dosen G"],
    "Dosen A": ["Dosen B"],
    "Dosen B": ["Dosen C"],


    "Dosen C": [],
    "Dosen D": [],
    "Dosen E": [],
    "Dosen F": [],
    "Dosen G": []
}

def bfs(start):
    visited = []
    queue = deque([start])

    while queue:
        node = queue.popleft()

        if node not in visited:
            visited.append(node)
            for neighbor in graph[node]:
                queue.append(neighbor)

    return visited



hasil_bfs = bfs("Rektor")
print("Urutan BFS:")
print(" → ".join(hasil_bfs))
```

```python
# IMPLEMENTASI ALGORITMA DEPTH-FIRST SEARCH

# DEPTH-FIRST SEARCH
def dfs(visited, graph, node):
    if node not in visited:
        print(node)
        visited.add(node)
        for neighbor in graph[node]:
            dfs(visited, graph, neighbor)
```

```python
# Latihan 2

graph = {
    "Rektor": ["Warek 1", "Warek 2"],
    "Warek 1": ["Kaprodi 1"],
    "Warek 2": ["Kaprodi 2", "Kaprodi 3"],
    "Kaprodi 1": ["Dosen A"],
    "Kaprodi 2": ["Dosen D", "Dosen E"],
    "Kaprodi 3": ["Dosen F", "Dosen G"],
    "Dosen A": ["Dosen B"],
    "Dosen B": ["Dosen C"],
    "Dosen C": [],
    "Dosen D": [],
    "Dosen E": [],
    "Dosen F": [],
    "Dosen G": []
}

def dfs(start):
    visited = []
    stack = [start]

    while stack:
        node = stack.pop()

        if node not in visited:
            visited.append(node)

            for neighbor in reversed(graph[node]):
                stack.append(neighbor)

    return visited


hasil_dfs = dfs("Rektor")
print("Urutan DFS:")
print(" → ".join(hasil_dfs))
```
